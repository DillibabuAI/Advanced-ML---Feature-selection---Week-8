```python
import pickle
import numpy as np
import pandas as pd

# =====================================================================
# 1. LOAD THE SAVED OBJECTS
# =====================================================================
# Load the pre-trained StandardScaler
with open("scaler_startups.sav", "rb") as scaler_file:
    sc = pickle.load(scaler_file)

# Load the pre-trained SVR Model
with open("finalized_model_svr.sav", "rb") as model_file:
    loaded_model = pickle.load(model_file)


# =====================================================================
# 2. DEFINE THE NEW DATA INSTANCE
# =====================================================================
# Define the precise feature columns your model expects
feature_cols = ['R&D Spend', 'Administration', 'Marketing Spend', 'State_Florida', 'State_New York']

# Example raw inputs for a new startup: 
# R&D Spend = 165349.20, Admin = 136897.80, Marketing = 471784.10, State = New York [0, 1]
raw_input_data = [[165349.20, 136897.80, 471784.10, 0, 1]]

# Convert to a DataFrame with column names to prevent UserWarnings
new_data_df = pd.DataFrame(raw_input_data, columns=feature_cols)


# =====================================================================
# 3. PRE-PROCESS AND PREDICT
# =====================================================================
# Scale the input data using the loaded scaler
preinput_scaled = sc.transform(new_data_df)

# Re-apply column headers to the scaled data for the model
preinput_scaled_df = pd.DataFrame(preinput_scaled, columns=feature_cols)

# Generate the final prediction
result = loaded_model.predict(preinput_scaled_df)

print("Predicted Profit Value:", result[0])
```
