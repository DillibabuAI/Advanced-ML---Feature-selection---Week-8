```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Importing the dataset
from sklearn.datasets import load_breast_cancer
```

```python
# Loading the data
cancer = load_breast_cancer()
```

```python
# Display the keys for the dataset
cancer.keys()
```

```python
# Describe about the dataset
print(cancer['DESCR'])
```

```python
# Display the dataset
df = pd.DataFrame(cancer['data'], columns = cancer['feature_names'])
df
```

```python
# Importing the StandardScaler
from sklearn.preprocessing import StandardScaler
```

```python
# Fitting the scaler model
sc = StandardScaler()
sc.fit(df)
```

```python
# Transforming the data
scaled_data = sc.transform(df)
```

```python
# Displaying the scaled data
scaled_data
```

```python
# Importng the PCA library
from sklearn.decomposition import PCA
```

```python
# Implementing the PCA method
pca = PCA(n_components = 2)
pca.fit(scaled_data)
```

```python
# Transforming the pca data
x_pca = pca.transform(scaled_data)
```

```python
# Display the shape of scaled data
scaled_data.shape
```

```python
# Display the pca converted data shape
x_pca.shape
```

```python
# Display scaled_data
scaled_data
```

```python
# Display pca data
x_pca
```

```python
# Plotting the diagram
plt.figure(figsize = (8,6))
plt.scatter(x_pca[:,0],x_pca[:,1],c=cancer['target'])
plt.xlabel('First principle component')
plt.ylabel('Second principle component')
```
